```python
def convert_latitude(lat_str):
    # Check if the last character is 'N' or 'S'
    if lat_str.endswith('N'):
        # Strip the 'N' and convert to float
        return float(lat_str[:-1])
    elif lat_str.endswith('S'):
        # Strip the 'S', convert to float, and multiply by -1
        return -float(lat_str[:-1])
    else:
        raise ValueError("Latitude must end with 'N' or 'S'.")
```

```python
def convert_longitude(lat_str):
    # Check if the last character is 'N' or 'S'
    if lat_str.endswith('E'):
        # Strip the 'N' and convert to float
        return float(lat_str[:-1])
    elif lat_str.endswith('W'):
        # Strip the 'S', convert to float, and multiply by -1
        return -float(lat_str[:-1])
    else:
        raise ValueError("Longitude must end with 'N' or 'S'.")
```

```python
df1['Latitude'] = df1['Latitude (Deg)'].apply(convert_latitude)
df1['Longitude'] = df1['Longitude (Deg)'].apply(convert_longitude)
```

```python
import matplotlib.pyplot as plt

# Assuming df2 is your DataFrame and you want to create a pie chart for the 'recclass' column
value_counts = df2['recclass'].value_counts()

# Calculate the total count of 'recclass' values
total_count = len(df2['recclass'])

# Set the threshold for percentage (e.g., 1%)
threshold_percentage = 1

# Filter 'recclass' values that are above the threshold
filtered_value_counts = value_counts[value_counts / total_count * 100 > threshold_percentage]

# Group the remaining 'recclass' values as 'Other'
filtered_value_counts['Other'] = total_count - filtered_value_counts.sum()

# Create a pie chart
plt.figure(figsize=(8, 8))
filtered_value_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90, legend=False)

# Add labels and title
plt.title('Distribution of recclass (Above 1%)')
plt.ylabel('')

# Display the pie chart
plt.show()

# Print the counts of meteorites in each class
print("\nCounts of meteorites in each class:")
for class_name, count in filtered_value_counts.items():
    print(f"{class_name}: {count}")
```
